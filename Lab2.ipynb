```python
r_num = int(input("Enter your roll number : "))
L = []

if r_num == 0:
    L = [0]
else:
    while r_num > 0:
        digit = r_num % 10
        L.append(digit)
        r_num = r_num // 10
    L.reverse()  # Reverse to restore original left-to-right digit order
    L=[i*10 for i in L]
print(L)
```

```python
def reverse_tuple_to_list(tup):
    # Tuples are immutable, so their elements cannot be modified or reordered in place.
    return list(reversed(tup))


# Example usage:
my_tuple = (1, 2, 3, 4, 5)
result = reverse_tuple_to_list(my_tuple)

print("Original tuple:", my_tuple)
print("Reversed list:", result)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example input list


# Build the list of prime numbers using list comprehension
primes = [num for num in random_num if is_prime(num)]

# Output results
print("List of prime numbers:", primes)
print("Total prime count:", len(primes))
```

```python
import statistics
single_mode = statistics.mode(random_num)
print("Mode:", single_mode)  # Output: 4

# Multi-mode (handles ties)
all_modes = statistics.multimode(random_num)
print("All modes:", all_modes)  # Output: [4]
```
