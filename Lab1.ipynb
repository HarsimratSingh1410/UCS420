```python
def IsPrime(n):
	for i in range(2, n//2 + 1):
		if n%i==0:
			return 0
	return 1

print ("IsPrime(21)  --> ", IsPrime(21))
print ("IsPrime(237)  --> ", IsPrime(237))
```

```python
var =" Indian   Army    "

print("String    --> ", var)
print("Length    --> ", len(var))
print("var strip --> ", var.strip())
print("Length of var after strip --> ", len(var.strip()))
```

```python
var =" Indian,   Army    "

print("String    :", var)
print("Length    : ", len(var))
print("var split :", var.split())
print("var split : ", var.split(' '))
print("var split : ", var.split(','))

# Strip + Split
print("var split --> ", var.strip().split(','))
```

```python
var=" Indian Army    "
print ("String       --> ", var)
print ("Count of ' ' --> ", var.count(' '))
print ("Count of 'a' --> ", var.count('a'))
print ("Count of 'n' --> ", var.count('an'))
```

```python
var="Indian Army"
print ("String    --> ", var)
print ("var[::1]  --> ", var[::1])
print ("var[::2]  --> ", var[::2])
print ("var[::-1] --> ", var[::-1])
print ("var[::-2] --> ", var[::-2])

var=var[::-1]
print ("var after reverse --> ", var)
```
